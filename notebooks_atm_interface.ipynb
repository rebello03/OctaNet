```python
import sys
import time


class ATM:

  def __init__(self, user_pin, initial_balance=1000.0):
    self.user_pin = user_pin
    self.balance = initial_balance
    self.transaction_history = []

  def verify_pin(self):
    attempts = 3
    while attempts > 0:
      entered_pin = input("\nEnter your 4-digit PIN: ").strip()
      if entered_pin == self.user_pin:
        print("PIN verified successfully!")
        return True
      else:
        attempts -= 1
        print(f"Incorrect PIN. {attempts} attempt(s) remaining.")
    print("\nToo many failed attempts. Your card is temporarily blocked.")
    return False

  def check_balance(self):
    print(f"\nCurrent Account Balance: ${self.balance:.2f}")
    self.transaction_history.append(
        f"Balance Inquiry: ${self.balance:.2f}"
    )

  def deposit(self):
    try:
      amount = float(input("\nEnter amount to deposit: $"))
      if amount <= 0:
        print("Invalid amount. Please enter a positive value.")
        return
      self.balance += amount
      print(f"${amount:.2f} deposited successfully.")
      print(f"New Balance: ${self.balance:.2f}")
      self.transaction_history.append(f"Deposited: +${amount:.2f}")
    except ValueError:
      print("Invalid input. Please enter a numeric value.")

  def withdraw(self):
    try:
      amount = float(input("\nEnter amount to withdraw: $"))
      if amount <= 0:
        print("Invalid amount. Please enter a positive value.")
      elif amount > self.balance:
        print("Insufficient funds! Transaction cancelled.")
      else:
        self.balance -= amount
        print(f"${amount:.2f} withdrawn successfully.")
        print(f"Remaining Balance: ${self.balance:.2f}")
        self.transaction_history.append(f"Withdrew: -${amount:.2f}")
    except ValueError:
      print("Invalid input. Please enter a numeric value.")

  def transfer(self):
    recipient = input("\nEnter recipient account number: ").strip()
    if not recipient.isdigit() or len(recipient) < 6:
      print("Invalid account number. Must be at least 6 digits.")
      return
    try:
      amount = float(input("Enter amount to transfer: $"))
      if amount <= 0:
        print("Invalid amount. Please enter a positive value.")
      elif amount > self.balance:
        print("Insufficient funds for transfer.")
      else:
        self.balance -= amount
        print(f"${amount:.2f} transferred to Account {recipient}.")
        print(f"Remaining Balance: ${self.balance:.2f}")
        self.transaction_history.append(
            f"Transferred: -${amount:.2f} to Acc #{recipient}"
        )
    except ValueError:
      print("Invalid input. Please enter a numeric value.")

  def view_history(self):
    print("\n--- Transaction History ---")
    if not self.transaction_history:
      print("No transactions recorded in this session.")
    else:
      for idx, transaction in enumerate(self.transaction_history, 1):
        print(f"{idx}. {transaction}")
    print("---------------------------")

  def change_pin(self):
    current = input("\nEnter current PIN: ").strip()
    if current != self.user_pin:
      print("Incorrect current PIN.")
      return
    new_pin = input("Enter new 4-digit PIN: ").strip()
    if len(new_pin) == 4 and new_pin.isdigit():
      self.user_pin = new_pin
      print("PIN updated successfully!")
      self.transaction_history.append("Security: PIN changed")
    else:
      print("Invalid format. PIN must be exactly 4 digits.")

  def run(self):
    print("====================================")
    print("      WELCOME TO OCTANET ATM        ")
    print("====================================")

    if not self.verify_pin():
      sys.exit()

    while True:
      print("\n========= MAIN MENU =========")
      print("1. Check Balance")
      print("2. Deposit Money")
      print("3. Withdraw Money")
      print("4. Transfer Funds")
      print("5. Transaction History")
      print("6. Change PIN")
      print("7. Exit")
      print("=============================")

      choice = input("Select an option (1-7): ").strip()

      if choice == "1":
        self.check_balance()
      elif choice == "2":
        self.deposit()
      elif choice == "3":
        self.withdraw()
      elif choice == "4":
        self.transfer()
      elif choice == "5":
        self.view_history()
      elif choice == "6":
        self.change_pin()
      elif choice == "7":
        print("\nThank you for using OctaNet ATM. Goodbye!")
        break
      else:
        print("Invalid choice. Please select a number between 1 and 7.")

      time.sleep(0.5)


if __name__ == "__main__":
  # Default PIN is 1234, Starting Balance is $1000.00
  my_atm = ATM(user_pin="1234", initial_balance=1000.0)
  my_atm.run()
```
